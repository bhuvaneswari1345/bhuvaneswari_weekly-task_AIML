```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
data = {
    'PetAge_Year': [9.8, 4.5, 11.2, 13.4, 6.6, 3.7, 0.9, 9.9, 3.7, 12.2],
    'PetWeight_Kg': [22.2, 11.3, 27.8, 30.3, 15.1, 10.7, 3.3, 24.5, 11.0, 27.4]
}
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
x = df[['PetAge_Year']]
y = df['PetWeight_Kg']
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
print(f"Intercept (c): {model.intercept_:.2f}")
print(f"Coefficient/Slope (m): {model.coef_[0]:.2f}")
```

```python
y_pred = model.predict(X_test)
```

```python
plt.scatter(x, y, color="blue", label="Actual Data")
plt.plot(x, model.predict(x), color="red", label="Regression Line")
plt.xlabel("Pet Age (Years)")
plt.ylabel("Pet Weight (Kg)")
plt.title("Pet Weight Prediction using Linear Regression")
plt.legend()
plt.show()
```
